```python
import tensorflow as tf
from tensorflow.keras.layers import (
    Input, Conv2DTranspose, Conv2D, MaxPooling2D
)
from tensorflow.keras.models import Model

inputs = Input(shape=(28, 28, 1))

# Reduce spatial dimensions
x = Conv2D(32, 3, activation="relu", padding="same")(inputs)
x = MaxPooling2D(pool_size=2)(x)  # 28x28 -> 14x14
x = Conv2D(64, 3, activation="relu", padding="same")(x)
x = MaxPooling2D(pool_size=2)(x)  # 14x14 -> 7x7

# Expand spatial dimensions using transpose convolution
x = Conv2DTranspose(
    64, 3, strides=2, padding="same", activation="relu"
)(x)  # 7x7 -> 14x14

x = Conv2DTranspose(
    32, 3, strides=2, padding="same", activation="relu"
)(x)  # 14x14 -> 28x28

outputs = Conv2D(1, 3, activation="sigmoid", padding="same")(x)

model = Model(inputs, outputs)
model.summary()
```
